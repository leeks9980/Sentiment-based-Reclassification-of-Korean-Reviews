```python
import requests
import time

def Collect_Steam_Reviews(APPID, game_id, TARGET_COUNT=1000, LANG="korean"):
    BASE_URL = f"https://store.steampowered.com/appreviews/{APPID}"
    params = {
        "json": 1,
        "language": LANG,
        "filter": "funny",
        "num_per_page": 100,
        "cursor": "*"
    }
    
    all_reviews = []
    
    print(f"--- Collecting reviews for {game_id} (Target: {TARGET_COUNT}) ---")

    while True:
        if len(all_reviews) >= TARGET_COUNT:
            break

        try:
            res = requests.get(BASE_URL, params=params)
            if res.status_code != 200:
                break
            
            data = res.json()
        except:
            break

        reviews = data.get("reviews", [])

        if not reviews:
            break

        for r in reviews:
            if len(all_reviews) >= TARGET_COUNT:
                break

            text = r["review"].replace("\n", " ").strip()
            
            if text == "":
                continue

            is_recommended = "Recommended" if r["voted_up"] else "Not Recommended"

            all_reviews.append({
                "game_id": game_id,
                "review": text,
                "votes_funny": r.get("votes_funny", 0),
                "votes_up": r["votes_up"],
                "voted_up": r["voted_up"],
                "is_recommended": is_recommended,
                "playtime_forever": r["author"]["playtime_forever"],
                "timestamp_created": r["timestamp_created"]
            })

        if params["cursor"] == data["cursor"]:
            break
            
        params["cursor"] = data["cursor"]
        time.sleep(0.3)

    return all_reviews
```

```python
a = []
for game_id_1, game_name in game_id_list.items():
    b = Collect_Steam_Reviews(game_id = game_name, APPID = game_id_1, TARGET_COUNT=1000)
    for i in b:
        a.append(i)       
```

```python
df = pd.read_csv("steam_reviews.csv")

def clean_review(text):
    if not isinstance(text, str):
        return ""
    
    # 줄바꿈 및 탭 제거
    text = text.replace("\n", " ").replace("\t", " ")

    # URL 제거
    text = re.sub(r"http\S+|www\S+", "", text)

    # HTML 태그 제거
    text = re.sub(r"<.*?>", "", text)

    # 4) 이모지 제거
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags
                               "]+", flags=re.UNICODE)
    text = emoji_pattern.sub(r"", text)

    # 특수문자 제거 (한글/영어/숫자 제외)
    text = re.sub(r"[^가-힣a-zA-Z0-9.,!? ]", " ", text)

    # 다중 공백을 하나로
    text = re.sub(r"\s+", " ", text).strip()

    return text

df["review"] = df["review"].apply(clean_review)

df = df.drop_duplicates(subset=["review"])

df.to_csv("steam_reviews_clean.csv", index=False, encoding="utf-8-sig")
```
